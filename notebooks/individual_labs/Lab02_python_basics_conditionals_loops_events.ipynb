# Part A (Week 2): Python Basics with Weather Variables

Part A works with one table of daily weather observations. The goal is to move
values out of the table into ordinary Python variables, lists, and
dictionaries, and to keep the units attached at every step. Finish Part A,
including Graded Checkpoint A, before the Week 3 session.

```python
import pandas as pd

weather = pd.read_csv(PROCESSED_DIR / "iowa_city_weather_daily.csv", parse_dates=["date"])
first_row = weather.iloc[0]
temp_c = first_row["temp_mean_c"]
temp_f = temp_c * 9 / 5 + 32
print(f"First daily mean temperature: {temp_c:.1f} deg C = {temp_f:.1f} deg F")
```

```python
weather_columns = ["temp_mean_c", "precipitation_mm", "wind_speed_mean_mps"]
units = {"temp_mean_c": "deg C", "precipitation_mm": "mm/day", "wind_speed_mean_mps": "m/s"}
for column in weather_columns:
    print(column, units[column], weather[column].mean())
```

```python
total_precip = weather["precipitation_mm"].sum()
wet_days = (weather["precipitation_mm"] > 1.0).sum()
print(f"Total precipitation: {total_precip:.1f} mm")
print(f"Number of days with more than 1 mm precipitation: {wet_days}")
```

## Guided coding: variables tell a story

Store single values in named variables, then combine them into a readable

```python
hottest = weather["temp_max_c"].max()
coldest = weather["temp_min_c"].min()
station_name = weather["station"].iloc[0]
report = f"Station {station_name}: extremes ranged from {coldest:.1f} to {hottest:.1f} deg C."
print(report)
```

## Guided coding: lists and indexing with real observations

A pandas column can be turned into a plain Python list. Lists support
positional indexing (`[0]`, `[-1]`) and slicing (`[1:4]`).

```python
first_week_temps = weather["temp_mean_c"].head(7).tolist()
print(first_week_temps)
print("First day:", first_week_temps[0])
print("Last day of the week:", first_week_temps[-1])
print("Days 2 to 4:", first_week_temps[1:4])

manual_mean = sum(first_week_temps) / len(first_week_temps)
print(f"Manual mean: {manual_mean:.2f} deg C")
print(f"pandas mean: {weather['temp_mean_c'].head(7).mean():.2f} deg C")
```

# Part B (Week 3): Conditionals, Loops, and Environmental Event Detection

Week 3 has no Monday lecture in this offering, because of the Labor Day

introduction to `if`/`elif`/`else` and `for` loops, then continues here and
through the Friday studio.

Part A turned table values into variables. Part B turns them into decisions. An
environmental event is a definition imposed on continuous data, so every count
below depends on a threshold that somebody chose and should be able to defend.

The weather table loaded in Part A is still in memory. The cell below adds the
daily streamflow record.

```python
# `weather` is already loaded in Part A. Add the daily streamflow record.
stream = pd.read_csv(PROCESSED_DIR / "iowa_streamflow_daily.csv", parse_dates=["date"])
print(f"Weather rows: {len(weather)}, streamflow rows: {len(stream)}")
```

```python
event_labels = []
for _, row in weather.iterrows():
    if row["precipitation_mm"] >= 10:
        label = "wet"
    elif row["temp_max_c"] >= 32:
        label = "hot"
    elif row["temp_min_c"] <= -10:
        label = "cold"
    else:
        label = "ordinary"
    event_labels.append(label)

weather["event_label"] = event_labels
print(weather["event_label"].value_counts())
```

```python
threshold = stream["discharge_cfs"].quantile(0.90)
stream["high_flow"] = stream["discharge_cfs"] >= threshold
stream["month"] = stream["date"].dt.month
monthly_high_flow = stream.groupby("month")["high_flow"].sum()
print(monthly_high_flow)
```

## Guided coding: how long was the longest dry spell?

Some questions need memory across loop steps. Here two counters track the
current run of dry days and the longest run seen so far.

```python
longest_dry = 0
current_dry = 0
for precip in weather["precipitation_mm"]:
    if precip < 1.0:
        current_dry = current_dry + 1
        if current_dry > longest_dry:
            longest_dry = current_dry
    else:
        current_dry = 0

print(f"Longest run of days with < 1 mm precipitation: {longest_dry} days")
```

## Guided coding: combined conditions, two ways

Muggy days need heat AND humidity at the same time. The loop version and the
vectorized version below must agree; comparing them is a free sanity check.

```python
muggy_loop = 0
for _, row in weather.iterrows():
    if row["temp_max_c"] >= 30 and row["relative_humidity_mean"] >= 70:
        muggy_loop = muggy_loop + 1

muggy_vectorized = int(((weather["temp_max_c"] >= 30) & (weather["relative_humidity_mean"] >= 70)).sum())
print("Muggy days, loop count:      ", muggy_loop)
print("Muggy days, vectorized count:", muggy_vectorized)
```

## Guided coding: a season function

Functions turn a rule into a reusable, testable unit. This one maps a month
number to a season name, and `.apply()` runs it on every row.

```python
def month_to_season(month):
    if month in (12, 1, 2):
        return "winter"
    elif month in (3, 4, 5):
        return "spring"
    elif month in (6, 7, 8):
        return "summer"
    else:
        return "fall"

weather["season"] = weather["date"].dt.month.apply(month_to_season)
wet_days_only = weather[weather["precipitation_mm"] >= 10]
print("Wet days (>= 10 mm) by season:")
print(wet_days_only["season"].value_counts())
```

## Common mistakes and debugging tips

- Check that `PROCESSED_DIR.exists()` printed `True`.
- Read error messages from the bottom upward.
- Check column names with `df.columns` before selecting a column.
- Keep units in figure labels and written interpretations.
- Re-run earlier cells after changing data-loading or helper-code cells.
- Part B depends on variables created in Part A, so run the notebook from the
  top after reopening it.
- In an `if`/`elif` chain only the first true branch runs. A day that is both
  wet and hot takes whichever label comes first.
- A counter inside a loop needs a reset path. Skipping the reset produces a
  wrong answer with no error message.
- Use `and`/`or` for plain Python conditions and `&`/`|` for pandas Series.
