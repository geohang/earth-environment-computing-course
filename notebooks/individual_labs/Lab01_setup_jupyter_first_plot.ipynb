## Required imports and project paths

Run this cell first. It finds the project root whether the notebook is opened from the
repository root or from a notebook folder.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "data").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_DIR = PROJECT_ROOT / "data"
PROCESSED_DIR = DATA_DIR / "processed"
SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

print(f"Project root: {PROJECT_ROOT}")
print(f"Processed data folder exists: {PROCESSED_DIR.exists()}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

weather = pd.read_csv(PROCESSED_DIR / "iowa_city_weather_daily.csv", parse_dates=["date"])
print(weather.head())
print(weather[["temp_mean_c", "temp_min_c", "temp_max_c"]].describe())
```

## Guided coding: first plot

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(weather["date"], weather["temp_mean_c"], color="firebrick", linewidth=0.8, label="Daily mean")
ax.set_title("Daily mean air temperature, Iowa City")
ax.set_xlabel("Date")
ax.set_ylabel("Temperature (deg C)")
ax.legend()
fig.tight_layout()
plt.show()
```

## Guided coding: know your table before you trust it

Before plotting anything, a careful scientist checks what the table actually
contains: how many rows, which columns, what data types, what time span, and
whether values are missing. Run the cell below and read every line of output.

```python
print("Rows and columns:", weather.shape)
print()
print(weather.dtypes)
print()
print("First date:", weather["date"].min())
print("Last date:", weather["date"].max())
print()
print("Missing values per column:")
print(weather.isna().sum())
```

## Guided coding: zoom into one summer

A boolean mask keeps only the rows where a condition is true. Here we keep
summer 2024 and plot the daily mean and daily maximum temperature together.

```python
summer_2024 = weather[(weather["date"] >= "2024-06-01") & (weather["date"] <= "2024-08-31")]
print(f"Days in the window: {len(summer_2024)}")
print(f"Summer 2024 mean temperature: {summer_2024['temp_mean_c'].mean():.1f} deg C")

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(summer_2024["date"], summer_2024["temp_mean_c"], color="firebrick", label="Daily mean")
ax.plot(summer_2024["date"], summer_2024["temp_max_c"], color="darkorange", linewidth=0.8, label="Daily max")
ax.set_title("Iowa City temperatures, summer 2024")
ax.set_xlabel("Date")
ax.set_ylabel("Temperature (deg C)")
ax.legend()
fig.tight_layout()
plt.show()
```

## Guided coding: find a record-setting day

`idxmin()` and `idxmax()` return the row label of the smallest or largest
value, and `.loc[...]` pulls out that whole row.

```python
coldest_index = weather["temp_min_c"].idxmin()
coldest_day = weather.loc[coldest_index]
print("Coldest day on record:", coldest_day["date"].date(), f"at {coldest_day['temp_min_c']:.1f} deg C")

hottest_index = weather["temp_max_c"].idxmax()
hottest_day = weather.loc[hottest_index]
print("Hottest day on record:", hottest_day["date"].date(), f"at {hottest_day['temp_max_c']:.1f} deg C")
```
